```python
dbutils.widgets.text("catalog", "default")
catalog = dbutils.widgets.get("catalog")
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog || '.bronze');
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog || '.silver');
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog || '.gold');
```

```python
from pyspark.sql.functions import *
```

```python
def add_metadata(df):
    df1 = df.withColumn("file_name", col("_metadata.file_name")).withColumn("file_path", col("_metadata.file_path")).withColumn("ingestion_time", current_timestamp())
    return df1
```

```python
path = "/Volumes/dev/bronze/raw/"
```

```python
df=spark.read.csv(
    f"{path}sales.csv",
    header=True,
    inferSchema=True
)

df1=add_metadata(df)

df1.write.mode("overwrite").saveAsTable(f"{catalog}.bronze.sales_job")
```
