```python
%sql
CREATE OR REPLACE VIEW IDENTIFIER(:catalog ||'.gold.top_5_customers_job') AS
SELECT customer_id, SUM(total_amount) AS total 
FROM IDENTIFIER(:catalog ||'.silver.sales_cleaned') 
GROUP BY customer_id 
ORDER BY total DESC LIMIT 5
```
