```python
dbutils.widgets.text("catalog", "default")
catalog = dbutils.widgets.get("catalog")
```

```python
sales_df = spark.read.table(f"{catalog}.bronze.sales_job")
```

```python
sales_cleaned_df = sales_df.dropDuplicates().dropna(how="all").drop("_rescued_data", "file_name", "file_path", "ingestion_time")
```

```python
sales_cleaned_df.write.mode("overwrite").saveAsTable(f"{catalog}.silver.sales_cleaned_job")
```
